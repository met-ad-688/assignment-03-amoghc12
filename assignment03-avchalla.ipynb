```python
import os, sys
# make sure Spark uses the same interpreter as your kernel
os.environ["PYSPARK_PYTHON"] = sys.executable
# (not always necessary, but avoids mismatched interpreters)

# ---------------------------------------
# 1. Import all required Python libraries
# ---------------------------------------
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
import plotly.io as pio

from pyspark.sql import SparkSession
from pyspark.sql import functions as F
from pyspark.sql.functions import col, when, regexp_replace

# Set random seed for reproducibility (as required)
np.random.seed(42)

# ---------------------------------------
# 2. Start a Spark session
# ---------------------------------------
spark = SparkSession.builder.appName("LightcastData").getOrCreate()

# ---------------------------------------
# 3. Load the dataset (CSV file must be in your 'data' folder)
# ---------------------------------------
file_path = "./data/lightcast_job_postings.csv"  # adjust if you saved with a different name
df = spark.read.option("header", "true") \
               .option("inferSchema", "true") \
               .option("multiLine", "true") \
               .option("escape", "\"") \
               .csv(file_path)

# Register as a SQL view (this is helpful for Spark SQL queries later)
df.createOrReplaceTempView("job_postings")

# ---------------------------------------
# 4. Optional: Preview schema and data (remove or comment out before final submission)
# ---------------------------------------
print("✅ Spark session started and dataset loaded successfully.")
df.printSchema()
df.show(5)
```
